```python
# Cell 7
# Q7

import numpy as np

def manhattan_distance(x1, y1, x2, y2):
    return abs(x2 - x1) + abs(y2 - y1)

def euclidean_distance(x1, y1, x2, y2):
    return np.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print("Manhattan Distance =",
      manhattan_distance(1, 2, 4, 6))

print("Euclidean Distance =",
      euclidean_distance(1, 2, 4, 6))
```

```python
# Cell 13
# Part 2 - Q5

import numpy as np

x = np.array([2, 0, 1, 5, 4, 1, 9])

A = np.array([[3, 8, 1],
              [6, 2, 9]])

indices = np.argsort(x)

print("Indices after sorting =", indices)

print("Sorted x =", x[indices])

min_location = np.unravel_index(A.argmin(), A.shape)

print("Minimum value location =", min_location)

print("Indices where x > 3 =", np.where(x > 3))

parts = np.split(x, [2, 5])

print("First part =", parts[0])
print("Second part =", parts[1])
print("Third part =", parts[2])

x_expanded = np.expand_dims(x, axis=0)

print("Expanded shape =\n", x_expanded)

x_squeezed = np.squeeze(x_expanded)

print("Squeezed array =", x_squeezed)
```
